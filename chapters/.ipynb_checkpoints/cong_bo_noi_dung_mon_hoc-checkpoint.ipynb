# Cập nhật nội dung các môn học và công bố
<hr>

Tình trạng: **Đúng tiến độ**

## Mô tả công việc

Rà soát tổng thể các môn học thuộc 12 ngành đào tạo/chương trình đào tạo của Trường Đại học Kinh tế Quốc dân Công nghệ. Đánh giá nội dung các môn học, loại bỏ các môn học trùng lặp.

## Các công việc đã làm

- Đã hoàn thiện phần mềm quản lý môn học
- Đã hoàn thiện việc nhập dữ liệu môn học của các chương trình đào tạo tại Trường Đại học Kinh tế Quốc dân lên phần mềm.
- Đã tổ chức dữ liệu môn học dưới dạng có cấu trúc và dễ dàng quản lý

## Những công việc cần hoàn thiện

Các đơn vị trong Trường Đại học Kinh tế Quốc dân cập nhật nội dung môn học.
